### 1) Plot $I$ and $P_{L}$ as a function of $V_{L}$

Create an array of the load voltage to be used in the calculations

```python
Vl = np.array([i * 0.1 for i in range(11)])  # TODO: create an array of the load voltages over the range 0 to 1.0 pu
```

Calculate the current and power in the load for each value of $V_{L}$

```python
#TODO: calculate the current and power in the load.
# Calculate load current
I = (E - Vl) / Rs

# Calculate load power
PL = Vl * I
```

Store the results in a pandas dataframe for easy analysis and plotting

```python
results = pd.DataFrame({'VL': Vl, 'I': I, 'PL': PL})
```

```python
# TODO: plot the results

fig, ax1 = plt.subplots()

# Current on left y-axis
ax1.plot(results['VL'], results['I'], 'b-o', label='Current')
ax1.set_xlabel('Load Voltage (pu)')
ax1.set_ylabel('Load Current (pu)', color='b')
ax1.tick_params(axis='y', labelcolor='b')

# Power on right y-axis
ax2 = ax1.twinx()
ax2.plot(results['VL'], results['PL'], 'r-o', label='Power')
ax2.set_ylabel('Load Power (pu)', color='r')
ax2.tick_params(axis='y', labelcolor='r')

# Formatting
plt.title('DC Power Flow Results')
ax1.grid(True)
plt.show()
```
