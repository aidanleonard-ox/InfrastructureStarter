# Circuit Analysis

```python
import numpy as np
```

```python
# initialize the variables
n = 3 # number of nodes
A = np.zeros((n,n)) # TODO: initialize the conductance matrix as zeros(n, n)
b = np.zeros(n) # TODO: initialize the source vector as zeros(n)
Rl = 0.01
Rh = 10
V = 1
```

## Full Solution

```python
# iterate over the rows and columns of the matrix A
# TODO: hint: use a nested loop
        # fill in the diagonal elements of A with the sum of the conductances (1/R)
        # TODO: hint: use an if statement to check for a diagonal
        
        # fill in the off-diagonal elements of A with the negative line conductances
        # TODO: hint: use an if statement to check if 1 away from the diagonal

for i in range(n):
    for j in range(n):
        if i == j:
            A[i, j] = 2/Rl + 1/Rh  # fill in the diagonal elements of A with the sum of the conductances (1/R)
        elif abs(i - j) == 1:
            A[i, j] = -1/Rl  # fill in the off-diagonal elements of A with the negative line conductances

# fill in the final diagonal element of A with the sum of the conductances (1/R) but only 1 line conductance

# Correct the final diagonal element
A[n-1, n-1] = 1/Rl + 1/Rh


# fill in the first element of the vector b with the voltage source divided by the line resistance
# TODO
b[0] = V / Rl

# solve the system of equations Ax = b for x
# TODO: hint: use np.linalg.solve(A, b) to solve for x
x = np.linalg.solve(A, b)

# Check
print("Conductance matrix A:")
print(A)

print("Source vector b:")
print(b)

print("Node voltages x:")
print(x)
```
